```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LogisticRegression
```

```python
df = pd.read_excel("C:/Users/sgour/Downloads/credit_risk_company_ratios (1).xlsx", sheet_name="Company Ratios", header=2)
df
```

```python
df_model = df[df["Company"] != "Yes Bank"].copy()

features = ["Current Ratio", "Debt-to-Equity", "Interest Coverage Ratio", "ROE (%)", "ROA (%)", "Operating Profit Margin (%)"]
X = df_model[features]
y = df_model["Risk Label (Y)"]
```

```python
X = X.copy()
X["Debt-to-Equity"] = X["Debt-to-Equity"].fillna(10.0)   # negative equity = worse than any real D/E, so we cap it high
X["ROE (%)"] = X["ROE (%)"].fillna(-50.0)                 # negative equity companies get a strongly negative ROE proxy
```

```python
y_binary = y.map({"High Risk": 1, "Low Risk": 0})
```

```python
model = LogisticRegression(max_iter=1000, C=0.1)
model.fit(X, y_binary)
```

```python
coef_table = pd.DataFrame({
    "Ratio": features,
    "Coefficient": model.coef_[0]
})
coef_table["Odds Ratio"] = np.exp(coef_table["Coefficient"])
print("Intercept:", model.intercept_[0])
coef_table
```

```python
df_model["Predicted Risk Probability"] = model.predict_proba(X)[:, 1]
df_model[["Company", "Risk Label (Y)", "Predicted Risk Probability"]].sort_values("Predicted Risk Probability", ascending=False)
```
